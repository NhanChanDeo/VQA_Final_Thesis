# Step 1: Create Hugging Face Dataset Repositories Programmatically

This notebook authenticates to Hugging Face Hub using the `HF_TOKEN` secret and automatically creates the target **dataset** repositories for:
1. **DocVQA Task 1 (SP-DocVQA):** `{username}/docvqa-task1-spdocvqa`
2. **TextVQA v0.5.1:** `{username}/textvqa-v0.5.1`

> **Prerequisite:** In Google Colab, click the **Key icon (Secrets)** in the left sidebar and add a secret named `HF_TOKEN` with **Write** permission.

```python
# 1. Install & import Hugging Face Hub
%pip install -q --upgrade huggingface_hub

import os
from huggingface_hub import HfApi, login

# Dual-environment Token Auth (Colab userdata + local env fallback)
try:
    from google.colab import userdata
    HF_TOKEN = userdata.get('HF_TOKEN')
    print("✓ Loaded HF_TOKEN from Colab Secrets.")
except (ImportError, ModuleNotFoundError):
    HF_TOKEN = os.getenv('HF_TOKEN')
    print("✓ Loaded HF_TOKEN from environment variables.")

if not HF_TOKEN:
    raise ValueError(
        "HF_TOKEN not found! In Colab, click the Key icon in the left sidebar "
        "and add 'HF_TOKEN' with Write access."
    )

login(token=HF_TOKEN)
api = HfApi(token=HF_TOKEN)
username = api.whoami()['name']
print(f"✓ Authenticated to Hugging Face Hub as user: {username}")
```

```python
# 2. Define target dataset repositories
DOCVQA_REPO = f"{username}/docvqa-task1-spdocvqa"
TEXTVQA_REPO = f"{username}/textvqa-v0.5.1"

target_repos = [DOCVQA_REPO, TEXTVQA_REPO]
print("Target Repositories to provision:")
for r in target_repos:
    print(f" - {r}")
```

```python
# 3. Create dataset repositories with exist_ok check
for repo_id in target_repos:
    if api.repo_exists(repo_id=repo_id, repo_type="dataset"):
        print(f"ℹ Repository already exists: https://huggingface.co/datasets/{repo_id}")
    else:
        repo_url = api.create_repo(
            repo_id=repo_id,
            repo_type="dataset",  # Must be 'dataset'
            exist_ok=True,
            private=False         # Set to True if you wish to make it private
        )
        print(f"✓ Successfully created repository: https://huggingface.co/datasets/{repo_id}")

print("\n✓ Step 1 Complete: Both dataset repositories are ready on Hugging Face Hub!")
```
