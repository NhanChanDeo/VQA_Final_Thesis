```python
import os
from huggingface_hub import HfApi

# 1. Khởi tạo API với Token của bạn
HF_TOKEN = os.getenv("HF_TOKEN")  # hoặc userdata.get('HF_TOKEN') trên Colab
api = HfApi(token=HF_TOKEN)

repos = ["gamusa/textvqa-canonical", "gamusa/docvqa-canonical"]

# 2. Tạo branch 'lossless' từ commit hiện tại của nhánh 'main'
for repo_id in repos:
  try:
    api.create_branch(
        repo_id=repo_id,
        repo_type="dataset",
        branch="lossless",
        exist_ok=True,  # Không báo lỗi nếu branch đã tồn tại
    )
    print(f"✓ Đã tạo thành công branch 'lossless' cho: {repo_id}")
  except Exception as e:
    print(f"✗ Lỗi khi tạo branch cho {repo_id}: {e}")
```
